```python
#declaring feature vector and target variable
x = df

y = df['status_type']
```

```python
#converting categorical into numerical data by  label encoding
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

x['status_type'] = le.fit_transform(x['status_type'])

y = le.transform(y)
```

```python
# summary of x
x.info()
```

```python
x.head()
```

```python
cols = x.columns
from sklearn.preprocessing import MinMaxScaler

ms = MinMaxScaler()

x= ms.fit_transform(x)
x= pd.DataFrame(x, columns=[cols])
```

```python
x.head()
```

```python
#Kmeans model with two clusters
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=0) 

kmeans.fit(x)
```

```python
df.isnull().sum()
```

```python
#k-means parameter #fixing the centroid
kmeans.cluster_centers_
```

```python
#finding inertia

kmeans.inertia_
```

```python
#Check quality of  classification  by the model

labels = kmeans.labels_

# check how many of the samples were correctly labeled
correct_labels = sum(y == labels)

print("Result: %d out of %d samples were correctly labeled." % (correct_labels, y.size))
```

```python
print('Accuracy score: {0:0.2f}'. format(correct_labels/float(y.size)))
# it is a weakly classified data
```

```python
#Use elbow method to find optimal number of clusters
from sklearn.cluster import KMeans
cs = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0)
    kmeans.fit(x)
    cs.append(kmeans.inertia_)
plt.plot(range(1, 11), cs)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('CS')
plt.show()
```

```python
#By the above plot, we can see that there is a kink at k=2.

#Hence k=2 can be considered a good number of the cluster to cluster this data.

#But, we have seen that I have achieved a weak classification accuracy of 1% with k=2.

#it is aweakly classified data wity an accuracy of 1%
```
